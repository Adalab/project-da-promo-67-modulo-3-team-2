# Análisis de los gráficos

## Variables numéricas

Análisis introductorio de las variables:

- El 75% de los empleados vive a 14km o menos del trabajo. La mayoría (moda) vive a 2 km de distancia, aunque la media es de 9,1km.

- El 50% de los empleados se encuentra entre un nivel (JobLevel) 1 y 3, siendo el más frecuente el nivel 1. 

- La mayoría de empleados ha sido evaluado con un 3 en una escala del 1 al 4. Por lo que podemos decir que la evaluación de la empresa hacia sus empleados es, en líneas generales, positiva o neutral, pero no negativa.

- El 50% de empleados recibió entre 2 y 3 formaciones el año pasado. La mayoría recibió 3.

- La edad promedio de los empleados es 37. El 50% de los empleados tiene entre 31 y 42 años.

- El 50% de empleados tienen entre un nivel 2 y un nivel 4 de educación, siendo el 3 el más frecuente.

- De media, los empleados han trabajado en 2 empresas a lo largo de su carrera, sin embargo, lo más frecuente es que solo hayan trabajado en 1.

- De acuerdo con el rango de edad de los trabajadores, de media llevan trabjando 11 años. La vida laboral del 50% de ellos oscila entre los 5 y los 15 años, siendo 10 el número de años más frecuente.

- Sobre la vida laboral dentro de la empresa, lo más frecuente es que los empleados tengan 5 años de antigüedad, y el promedio se sitúa en los 7 años, ya que hay trabajadores que llevan en la compañía más de 20 años, que suben la media.

- Lo que también se observa es que los empleados con más antigüedad acumulan más años sin ascensos, por lo que puede haber un estancamiento laboral de los más veteranos.

- En cuanto a los salarios, el 50% de ellos se sitúa en los valores más bajos del rango, con presencia de valores atípicamente altos. El salario mediano y también el más frecuente son 4907. Lo que se aprecia es una diferencia muy amplia entre los salarios más bajos y los más altos, con concentración en los bajos.

- Sobre los aumentos salariales, la distribución es muy similar, es decir, el 50% de los porcentajes de aumento se concentran en los valores más bajos, entre el 12 y el 18%, siendo un 11% el porcentaje de aumento más frecuente. Llama la atención la diferencia con respecto al promedio, de 15,21%, lo que indica que hay valores más altos de aumento salarial que suben la media.

#### Análisis encuestas


De las encuestas realizadas a los empleados, que puntúan a la empresa del 1 al 4 en diferentes ámbitos, podemos deducir que:

- En las cinco encuestas, la respuesta más frecuente es 3, por lo que en líneas generales, podemos decir de que hay una valoración neutra/positiva de la empresa por parte de los trabajadores. .

- En relación con la satisfacción con el ambiente laboral, el 50% de los valores se encuentra entre 2 y 4, por lo que hay una parte importante de los votos que se encuentra en los valores más altos. Esta Distribución se comparte también la satisfacción con el trabajo

- En el compromiso con el trabajo, sin embargo, aunque la media sigue en 2.73 y la moda en 3, el 50% de los valores se concentra entre 2 y 3. Lo mismo sucede con la puntuación de equilibrio entre la vida personal y laboral.

- Por tanto, hay una mayor satisfacción general con el ambiente laboral y con el trabajo, mientras que las puntuaciones de compromiso y equilibrio entre vida laboral y personal son más neutrales que positivas. Si bien, en ninguno de estos ámbitos se aprecia una mala puntuación generalizada.

## Variables categóricas

**Análisis de las variables categóricas:**

- La mayoría del personal (83.88%) sigue en la empresa, por lo que, a priori, no parece que haya un problema grave de rotación de personal.

- Más de la mitad (64.69%) de los trabajadores viajan ocasionalmente.

- No hay un número parejo en los diferentes departamentos,sino que se observa que hay algunos con más trabajadores que otros. `Research & Development` (63.88%) y `Sales` (29.86%) son los que más plantilla tienen.

- La mayoría del personal tienen estudios relativos a `Life Sciences` (39.52%) y `Medical` (30.41%).

- En cuanto a la distribución por sexo de la plantilla, sí se observa un mayor porcentaje de hombres que de mujeres (60% y 40% respectivamente)

- Los puestos con mayor porcentaje son: `Sales Executive`, `Research Scientist`, `Laboratory Technician`. Representantes y directores hay menos, por lo que no es una empresa que jerárquicamente tenga una pirámide invertida. 

- En cuanto al estado civil, hay un reparto escalonado, con un 41.22% casados, un 29% solteros y un 20% divorciados.

- La gran mayoría de los trabajadores no realizan horas extra (72.45%).
