# Análisis de los gráficos

## Variables numéricas

Análisis introductorio de las variables:

- El 75% de los empleados vive a 14km o menos del trabajo. La mayoría (moda) vive a 2 km de distancia, aunque la media es de 9,1km.

- El 50% de los empleados se encuentra entre un nivel (JobLevel) 1 y 3.

- La mayoría de empleados ha sido evaluado con un 3 en una escala del 1 al 4. Por lo que podemos decir que la evaluación de la empresa hacia sus empleados es, en líneas generales, positiva o neutral, pero no negativa.

- El 50% de empleados recibió entre 2 y 3 formaciones el año pasado. La mayoría recibió 3.
