## ¿Qué factores influyen en la rotación del talento?

### Salario, porcentaje de aumento salarial y nivel y puesto de trabajo

De acuerdo con el análisis de `MonthlyIncome`, la rotación parece ser más elevada entre los empleados con rangos salariales más bajos. En el primer intervalo salarial, el porcentaje de salida alcanza el 26,8%, claramente por encima de la mayoría de tramos posteriores. Aunque existen algunas excepciones, como el intervalo salarial medio de 8.130 a 10.504, donde la rotación vuelve a aumentar hasta el 18,4%, en términos generales los porcentajes de salida tienden a reducirse en los rangos salariales más altos.

Esta lectura debe interpretarse conjuntamente con `JobLevel`, ya que ambas variables presentan una correlación positiva fuerte: los niveles profesionales más bajos suelen estar asociados a salarios más bajos. En este sentido, resulta coherente que el nivel 1 concentre el mayor porcentaje de rotación, con un 26,3% de empleados que abandonan la empresa. Sin embargo, este dato tampoco permite afirmar que el salario o el nivel profesional sean factores determinantes por sí solos, ya que la mayoría de empleados del nivel 1 permanece en la compañía, con un 73,7%.

Por tanto, los datos sugieren que los perfiles de menor nivel y menor salario presentan un mayor riesgo de salida, aunque la rotación no parece explicarse únicamente por estas variables. Es probable que el efecto del salario y del nivel profesional se combine con otros factores, como la antigüedad en la empresa, el tiempo en el rol, las oportunidades de desarrollo o el encaje inicial con el puesto.

Esto explica también que el porcentaje de incremento salarial no muestre una relación clara con la rotación. Aunque existen diferencias entre intervalos, no se observa una tendencia progresiva que indique que menores subidas salariales estén asociadas sistemáticamente a una mayor salida de empleados. De hecho, algunos tramos con incrementos más altos también presentan porcentajes de rotación elevados. Por tanto, `PercentSalaryHike` no parece actuar como una variable especialmente determinante por sí sola.

#### Puesto de trabajo

El análisis por `JobRole` muestra diferencias relevantes en la rotación según el tipo de puesto. El porcentaje de salida es especialmente elevado entre los `Sales Representative`, donde el 39,8% de los empleados abandona la empresa. También destacan, aunque con menor intensidad, los perfiles de `Laboratory Technician` y `Human Resources`, con porcentajes de salida del 23,9% y 23,1%, respectivamente.

Este resultado complementa el análisis de `MonthlyIncome` y `JobLevel`. La rotación era más elevada entre los empleados con menor salario y menor nivel profesional, y el análisis por puesto permite concretar mejor dónde se concentra ese riesgo. En particular, los `Sales Representative` parecen ser el colectivo más vulnerable, ya que tras cruzar `JobRole` con `MonthlyIncome`, `JobLevel` y rotación, hemos comprobado que se trata de perfiles de menor nivel y menor retribución, a quienes siguen Laboratory Technician y Human Resources.

En cambio, los puestos de mayor responsabilidad o mayor especialización, como `Manager`, `Research Director` o `Manufacturing Director`, presentan porcentajes de salida mucho más bajos. Esto refuerza la idea de que la rotación no afecta por igual a toda la organización, sino que se concentra más en determinados perfiles profesionales, especialmente en puestos comerciales, técnicos o de menor nivel jerárquico.

Por tanto, el salario y el nivel profesional parecen estar relacionados con la rotación, pero el análisis por puesto muestra que no basta con mirar únicamente variables económicas o jerárquicas. El tipo de rol también introduce diferencias importantes y permite identificar colectivos concretos con mayor riesgo de salida.
Esto apunta a que la rotación está más vinculada a perfiles junior o de menor retribución que a puestos consolidados dentro de la estructura organizativa.

### Trayectoria interna: años en la empresa, años desde la última promoción, años en el mismo puesto y con el mismo manager

Al analizar el porcentaje de salida dentro de cada tramo, los datos no respaldan claramente la idea de que la rotación se deba principalmente a largos periodos sin ascenso o a muchos años de permanencia en el mismo puesto. En `YearsInCurrentRole` y `YearsWithCurrManager`, el porcentaje de salida es mayor entre quienes llevan menos tiempo en el rol actual o con su responsable, lo que apunta más a fases iniciales de adaptación al puesto, al equipo o al estilo de liderazgo que a un estancamiento prolongado.

En cambio, `YearsSinceLastPromotion` muestra una relación menos definida: algunos tramos con más años desde el último ascenso presentan porcentajes de salida elevados, pero no existe una tendencia progresiva clara. Por tanto, esta variable podría influir en determinados perfiles, aunque no parece determinante por sí sola.

En cuanto a `YearsAtCompany`, la rotación es más elevada entre empleados con menor antigüedad, especialmente en el primer tramo analizado. Los empleados con trayectorias intermedias presentan porcentajes de salida más bajos, lo que sugiere mayor estabilidad una vez superada la etapa inicial en la compañía. No obstante, los tramos de mayor antigüedad deben interpretarse con cautela, ya que pueden estar compuestos por pocos casos.

CONCLUSIÓN: En conjunto, estas variables apuntan a que la rotación está más vinculada a etapas tempranas de adaptación (a la empresa, al rol y al manager) que a un estancamiento profesional prolongado. No obstante, la falta de promoción reciente podría estar influyendo en perfiles concretos, aunque no aparece como un patrón general claramente determinante.

### Horas extra

La variable `OverTime` muestra una relación clara con la rotación. Entre los empleados que no realizan horas extra, el porcentaje de salida es del 10,5%, mientras que entre quienes sí hacen horas extra asciende al 30,9%. Esto sugiere que las horas extra pueden estar asociadas a un mayor riesgo de abandono. No obstante, no pueden considerarse un factor determinante por sí solo, ya que el 69,1% de los empleados que realizan horas extra permanece en la compañía.

Por tanto, las horas extra parecen actuar como un factor de riesgo relevante, posiblemente relacionado con carga de trabajo, desgaste o dificultad para conciliar. Sin embargo, su efecto debe interpretarse junto con otras variables, como salario, rol, antigüedad, satisfacción laboral o equilibrio vida-trabajo.

Aunque las horas extra están claramente asociadas con una mayor rotación, esta relación no parece explicarse únicamente por una mala valoración del equilibrio vida-trabajo. Entre los empleados que hacen horas extra y se marchan, el 33,6% presenta una puntuación baja en `WorkLifeBalance`, mientras que la mayoría, el 66,4%, valora esta dimensión con puntuaciones medias o altas. Por tanto, las horas extra podrían estar influyendo en la salida, pero no necesariamente a través de una percepción negativa explícita de la conciliación.

Al analizar las horas extra junto con `JobLevel`, se observa que las horas extra están bastante repartidas entre los distintos niveles, por lo que no se concentran en los niveles más bajos. Sin embargo, lo que sí se observa es que en el nivel 1, entre los que sí hacen horas extra, hay un 64,8% que decide marcharse. Por tanto, lo que sí podemos afirmar es que las horas extra son un factor de riesgo para aumentar la rotación entre los niveles profesionales inferiores.

### Encuestas de satisfacción a los empleados

Las variables que parecen más relevantes son ```JobInvolvement``` y ```WorkLifeBalance```. En ambos casos, las puntuaciones más bajas están asociadas a porcentajes de salida claramente superiores. En ```JobInvolvement```, además, la relación es bastante ordenada: a mayor implicación, menor rotación. En ```WorkLifeBalance```, la puntuación 1 destaca claramente como grupo de riesgo, con un 31,2% de salida.

También hay relación en ```EnvironmentSatisfaction``` y ```JobSatisfaction```, aunque de forma más moderada. En estas variables, la rotación es más alta cuando la puntuación es 1, pero las diferencias entre las puntuaciones 2, 3 y 4 no son tan grandes. Esto sugiere que el verdadero punto crítico no es tanto pasar de una valoración media a una alta, sino tener una valoración claramente negativa.

La variable con menor capacidad explicativa parece ser RelationshipSatisfaction, ya que solo la puntuación 1 se diferencia algo del resto, mientras que las demás categorías presentan porcentajes de salida muy parecidos.

CONCLUSIÓN: Las variables de satisfacción e implicación laboral muestran una relación con la rotación, especialmente cuando las puntuaciones son muy bajas. Los empleados que valoran peor su implicación en el trabajo, el equilibrio vida-trabajo, el entorno laboral o la satisfacción con el puesto presentan mayores porcentajes de salida. Sin embargo, la relación no es igual de intensa en todas las variables. ```JobInvolvement``` y ```WorkLifeBalance``` parecen tener una asociación más clara con la rotación, mientras que ```RelationshipSatisfaction``` muestra una relación más débil. En general, los datos sugieren que las valoraciones negativas pueden actuar como señales de riesgo, aunque no explican por sí solas la decisión de abandonar la empresa.

### Viajes por trabajo

La variable ```BusinessTravel``` muestra una relación clara con la rotación. Los empleados que no viajan por trabajo presentan el menor porcentaje de salida, con un 8,6%, mientras que entre quienes viajan frecuentemente la rotación asciende al 24,7%. La categoría `Travel Rarely` se sitúa en un punto intermedio, con un 14,4%.

No obstante, al cruzar esta variable con ```WorkLifeBalance```, ```JobInvolvement``` y ```JobSatisfaction```, no se observa que quienes viajan frecuentemente puntúen peor en estas dimensiones. Tampoco parece haber una diferencia salarial clara según la frecuencia de viajes, ya que las medianas de ```MonthlyIncome``` son similares entre los distintos grupos.

El cruce con ```JobRole``` sí aporta un matiz relevante: ```Sales Representative``` es el puesto con mayor proporción de viajes frecuentes, con un 27,7%, y también el que presenta mayor rotación, menor salario mediano y un JobLevel mediano de 1.

Por tanto, los viajes frecuentes parecen estar asociados a una mayor rotación, pero no explican por sí solos la salida de empleados. Su efecto parece más relevante cuando se combina con otros factores, especialmente en perfiles de menor nivel, menor salario y mayor exposición a desplazamientos, como ocurre con ```Sales Representative```.

## Conclusiones


1. La empresa no presenta, en términos generales, un problema grave de rotación, ya que la mayoría de empleados permanece en la organización (83.88%). La salida de talento se concentra especialmente en determinados grupos de empleados.

2. Los datos apuntan a que la rotación es mayor entre los perfiles de menor nivel profesional y menor retribución. Los empleados con salarios más bajos y `JobLevel` 1 presentan porcentajes de salida superiores al resto. Esta tendencia se confirma al analizar el puesto de trabajo: `Sales Representative` es el rol con mayor rotación, con un 39,8%, y además combina varios factores de riesgo, como menor salario mediano, menor nivel profesional y mayor exposición a viajes frecuentes. También destacan, aunque con menor intensidad, los perfiles de `Laboratory Technician` y `Human Resources`.

3. En cuanto a la trayectoria interna, la rotación parece estar más vinculada a fases iniciales de adaptación, ya que los porcentajes de salida son más elevados entre quienes llevan menos tiempo en la empresa, posiblemente por dificultades de encaje, expectativas no satisfechas o adaptación al rol. No obstante, también se observa que los empleados con mayor antigüedad acumulan más años sin promoción, lo que podría apuntar a situaciones concretas de estancamiento entre perfiles más veteranos.

4. Las horas extra aparecen como uno de los factores más claramente asociados a la rotación. En general, están bastante repartidas entre los diferentes puestos. Entre quienes no realizan horas extra, la salida es del 10,5%, mientras que entre quienes sí las realizan asciende al 30,9%. El efecto de las horas extra parece ser más relevante cuando se combina con otros factores, especialmente con niveles profesionales bajos.

5. Sobre las encuestas internas, las puntuaciones bajas en `JobInvolvement`, `WorkLifeBalance`, `EnvironmentSatisfaction` y `JobSatisfaction` se asocian con mayores porcentajes de salida. Entre ellas, `JobInvolvement` y `WorkLifeBalance` parecen tener una relación más clara con la rotación. Aun así, estas variables no explican por sí solas la salida de empleados, sino que funcionan mejor como señales de alerta cuando aparecen combinadas con otros factores de riesgo.

6. Los viajes frecuentes por trabajo también muestran una relación con la rotación. Los empleados que no viajan presentan el menor porcentaje de salida, mientras que quienes viajan frecuentemente alcanzan un porcentaje notablemente superior. El caso más relevante vuelve a ser `Sales Representative`, puesto que combina mayor frecuencia de viajes, menor salario, menor nivel profesional y mayor rotación.

En conjunto, la rotación parece responder a una combinación de factores más que a una única causa aislada. Los perfiles con mayor riesgo de salida son aquellos que concentran varias condiciones: menor nivel profesional, menor salario, etapas iniciales en la empresa o en el rol, realización de horas extra, viajes frecuentes y puntuaciones bajas en implicación o equilibrio vida-trabajo.

Por tanto, si la empresa quisiera reducir la rotación, debería prestar especial atención a los perfiles más junior o de menor retribución, especialmente en puestos como `Sales Representative`, `Laboratory Technician` y `Human Resources`. También sería recomendable revisar los procesos de incorporación y adaptación al puesto, las cargas de trabajo asociadas a horas extra, las oportunidades de desarrollo profesional y las condiciones específicas de los roles con mayor exposición a viajes o desgaste.
